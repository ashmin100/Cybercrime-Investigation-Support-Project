### ***JPG augmentation***
- If two .jpg files show the same scene (e.g. same photo or frame) but have different quality levels (different compression, resolution, or noise), then technically they are different data, even though they may look the same visually.
- So, changing jpg quality can be compression-based augmentation (or image-quality augmentation).

```python
import os
from PIL import Image

src_dir = r"data\20250920\extracted_frame" 
dst_prefix = r"data\20251013\augmented_frame"
quality_levels = [95, 85, 75, 65, 50, 35] 

for root, dirs, files in os.walk(src_dir):
    for fname in files:
        if not fname.lower().endswith((".jpg", ".jpeg", ".png")):
            continue

        src_path = os.path.join(root, fname)
        stem, _ = os.path.splitext(fname)
        
        img = Image.open(src_path).convert("RGB")

        dst_dir = root.replace(src_dir, dst_prefix)
        os.makedirs(dst_dir, exist_ok=True)

        for q in quality_levels:
            out_name = f"{stem}_q{q}.jpg"
            out_path = os.path.join(dst_dir, out_name)
            img.save(out_path, format="JPEG", quality=q, optimize=True)

        print(f"Generated qualities for: {out_path}")

    print("Done — all JPEG versions created.")
```
